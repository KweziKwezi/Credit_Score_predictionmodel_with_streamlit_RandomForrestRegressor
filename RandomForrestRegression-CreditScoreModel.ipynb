```python
import pandas as pd 
Credit_Data = pd.read_csv('loan_data.csv')
```

```python
Credit_Data.head()
```

```python
y = Credit_Data.credit_score 
features = ["person_age","person_gender","person_education","person_income","person_emp_exp","person_home_ownership","cb_person_cred_hist_length","previous_loan_defaults_on_file"]
x = pd.get_dummies(Credit_Data[features])
```

```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, mean_absolute_percentage_error
from sklearn.model_selection import train_test_split

initial_model = RandomForestRegressor(random_state =1)
train_x,val_x,train_y,val_y = train_test_split(x,y,random_state=1)
initial_model.fit(train_x,train_y)

prediction = initial_model.predict(val_x)

mae = mean_absolute_error(prediction,val_y)

mape = mean_absolute_percentage_error(prediction,val_y)

print(mae , mape)
```

```python
def tuning(max_leaf_node,train_x,val_x,train_y,val_y):
    predict = RandomForestRegressor(random_state=1,max_leaf_nodes=max_leaf_node)
    predict.fit(train_x,train_y)
    value = predict.predict(val_x)
    mae = mean_absolute_error(value,val_y)

    return mae

for max_leaf_node in [2,2,5,10,20,30,50,100,150,200,250,300,400,None]:
 mae = tuning(max_leaf_node,train_x,val_x,train_y,val_y)
 print(f"max_leaf_node: {max_leaf_node}\tMean absolute error: {mae}")
```

```python
def tuning(min_samples_leaf,train_x,val_x,train_y,val_y):
    predict = RandomForestRegressor(random_state=1,max_leaf_nodes=30, min_samples_leaf=min_samples_leaf)
    predict.fit(train_x,train_y)
    value = predict.predict(val_x)
    mae = mean_absolute_error(value,val_y)

    return mae

for min_samples_leaf in [2,3,4,5,6,7,8,9,10]:
 mae = tuning(min_samples_leaf,train_x,val_x,train_y,val_y)
 print(f"min_samples_leaf: {min_samples_leaf}\tMean absolute error: {mae}")
```
